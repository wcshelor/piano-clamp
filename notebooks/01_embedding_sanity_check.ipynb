# CLaMP 3 embedding sanity check

Read the external run selected by `CLAMP3_RUN_ROOT`; do not write generated artifacts into the repository. C2 score and prompt vectors belong to one space. SAAS audio vectors are intentionally excluded.

```python
import os
from pathlib import Path

import numpy as np
import pandas as pd

run_dir = Path(os.environ["CLAMP3_RUN_ROOT"]) / "chopin_mozart_v0"
similarities = pd.read_csv(run_dir / "tables" / "prompt_similarity.csv")
features = pd.read_csv(run_dir / "tables" / "musicxml_features.csv")
contrasts = pd.read_csv(run_dir / "tables" / "composer_prompt_contrasts.csv")
similarities.head()
```

```python
score_shapes = {path.stem: np.load(path, allow_pickle=False).shape for path in (run_dir / "embeddings" / "scores").glob("*.npy")}
assert score_shapes and set(score_shapes.values()) == {(1, 768)}
display(similarities.groupby(["composer", "prompt_id"])["cosine_similarity"].agg(["count", "mean", "std"]))
display(contrasts.loc[contrasts["is_primary"]])
display(features.groupby("composer").mean(numeric_only=True))
```
